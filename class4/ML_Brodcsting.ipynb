```python
import numpy as np
```

```python
arr=np.arange(1,17).reshape(4,4)
print(arr)
arr1=np.arange(1,5)
print(arr1)
print(arr1+arr)
```

```python
row=np.arange(3,7)
print(row)
col=np.arange(3,7).reshape(4,1)
print(col)

print(col+row)
```

```python
arr3=np.arange(1,41).reshape(5,-1)
print(arr3,arr3.shape)
```

```python
arr=np.arange(1,13)
arr1=arr.reshape(1,-1)
arr2=arr.reshape(-1,1)
print("row",arr1)
print("col",arr2)
```

```python
arr=np.arange(1,13).reshape(4,3)
print(arr)
arr1=arr.ravel()#shellow copy 
print(arr1)
arr1[2]=100
print(arr1)
arr2=arr.flatten()#deep copy 
print(arr2)
arr2[2]=200
print(arr2)
```

```python
m=np.ones(3)
print(m)
add_arr=[10,20,30]
print(add_arr)

print(m+add_arr)
```

```python
table=np.arange(1,101).reshape(10,10)

print(table)
```
