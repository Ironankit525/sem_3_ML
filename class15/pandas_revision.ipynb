## 4. Cleaning

```python
# blanks per column
p.isnull().sum()
```

```python
# repeated rows
print(p.duplicated().sum())

# remove repeated rows
p_clean = df.drop_duplicates()

p.head()
```

```python
# remove rows where one column is blank
p_clean = p_clean.dropna(subset=['internships'])

# fill blanks with a value
p_clean['prep_hours'] = p_clean['prep_hours'].fillna(0)

# change the type
p_clean['internships'] = p_clean['internships'].astype(int)
```

```python
# new column from existing columns
p_clean['cgpa_percent'] = (p_clean['cgpa'] * 9.5).round(1)
p_clean[['cgpa', 'cgpa_percent']].head()
```
