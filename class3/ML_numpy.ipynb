```python
import numpy as np
```

```python
arr1=np.arange(10)
print(arr1)
arr1=arr1+10
print(arr1)
print("multi",arr1**5)
arr2=np.arange(30+2)
arr3=np.arange(30+2)
print(arr2+arr3)
```

```python
x=np.arange(10).reshape(5,2)
y=np.arange(10,20).reshape(5,2)
print(x**y)
print(x<y)
print(x>y)
print(x/y)
print(x-y)
```

```python
print(np.random.randn(10))
print(np.random.rand(10))
print(np.random.seed(90))
```

```python
print(np.random.randint(9,100,size=(2,3)))
print(np.random.randint(9,100,10))
```

```python
lst=[3,4,5,6,7,8,32,4,6,4563,356,5,563,65,35,65,4563,65,3]
print(np.random.choice(lst,2))
print(np.random.choice(3,9))
print(np.random.choice(lst,2,replace=False))
print(np.random.choice(10,5))
ro=np.arange(20)
np.random.shuffle(ro)
print(ro)
```

```python
st=np.arange(1,11).reshape(5,2)
print(st)
print(st.min())
print("row min",st.min(axis=1))
print("col min",st.min(axis=0))
print("total",st.sum())
print("total",st.sum(axis=1))
```
