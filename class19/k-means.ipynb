```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

```python
X,y=make_blobs(n_samples=100000,centers=5,random_state=50)

model=KMeans(n_clusters=5,random_state=50)
labels=model.fit_predict(X)


sns.scatterplot(x=X[:,0],y=X[:,1],hue=labels)
plt.show()
```

```python
X,y=make_blobs(n_samples=10000,centers=5,random_state=50)
wcss=[]



for i in range(1,20):
    model=KMeans(n_clusters=i,random_state=50)
    model.fit(X)
    wcss.append(model.inertia_)
    
    
sns.lineplot(x=range(1,20),y=wcss,marker='o')
plt.xlabel('no. of cluster')
plt.ylabel('wcss')
plt.show()

    
```

```python
from sklearn.datasets import make_blobs
from scipy.cluster.hierarchy import dendrogram,linkage
import matplotlib.pyplot as plt

X,y=make_blobs(n_samples=20,centers=3)

Z=linkage(X,method='ward')

plt.figure(figsize=(10,5))
dendrogram(Z)
plt.title("agglomerative clustring dendogram")
plt.xlabel("data poit")
plt.ylabel("distance")
plt.show()
```
