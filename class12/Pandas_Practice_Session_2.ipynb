# 🧮 **7. GroupBy: First/Last Values & size() vs count()**

### _`.first()` / `.last()` → first/last row's value in each group | `.size()` → number of rows per group (includes NaNs) | `.count()` → number of non-null values per group, per column_

---


#### **Q7.1 Group `df` by `Pclass` and use `.first()` and `.last()` to see the first and last passenger listed for each class.**

---

```python
print("First passenger listed in each class:")
display(df.groupby('Pclass').first())

print("Last passenger listed in each class:")
display(df.groupby('Pclass').last())

```

#### **Q7.2 Group `df` by `Embarked` and compare `.size()` vs `.count()`. Are the numbers the same for every column? Why might they differ for a column like `Age`?**

---

```python
print("size() -> rows per group:")
print(df.groupby('Embarked').size())

print()
print("count() -> non-null values per group, per column:")
display(df.groupby('Embarked').count())

# No, the numbers are not the same for every column.
# .size() counts every row in the group, including rows with NaN values.
# .count() only counts non-null values, so columns with missing data (Age, Cabin)
# show smaller numbers than .size(), while complete columns (Name, Fare) match it.
```

#### **Q7.3 Group `df` by `Pclass` and use `.size()` to count passengers per class, then use `.count()` on just the `Age` column to see how many passengers per class have a known age.**

---

```python
print("Passengers per class (size):")
print(df.groupby('Pclass').size())

print()
print("Passengers with a known Age per class (count):")
print(df.groupby('Pclass')['Age'].count())
```

# 🏷️ **8. Named Aggregations with agg()**

#### **Q8.1 Group `df` by `Pclass` and use named aggregation to build a summary with columns `avg_fare` (mean of Fare) and `max_age` (max of Age).**

# 🧩 **9. Custom Aggregations with agg()**

### _`.agg()` isn't limited to built-in strings like `'mean'` or `'max'` — you can also pass a `lambda` or your own custom function to compute anything you need per group_

---


#### **Q9.1 Group `df` by `Pclass` and use `.agg()` with a `lambda` function to find the range (`max - min`) of `Fare` for each class.**

---

```python
fare_range = df.groupby('Pclass')['Fare'].agg(lambda x: x.max() - x.min())
fare_range
```

#### **Q9.2 Write your own custom function called `age_range(x)` that returns `x.max() - x.min()`. Group `df` by `Embarked` and pass this function into `.agg()` to find the age range for each embarkation port.**

---

```python
def age_range(x):
    return x.max() - x.min()

df.groupby('Embarked')['Age'].agg(age_range)
```

# 🔍 **11. Filtering Groups**

### _`.groupby('col').filter(func)` → keep only the groups for which a condition on the WHOLE group is True — different from row-wise filtering_

---


#### **Q11.1 Group `df` by `Embarked` and keep only the groups that have more than 100 passengers, using `.groupby().filter()`.**

---

```python
big_ports = df.groupby('Embarked').filter(lambda g: len(g) > 100)

print("Passengers per port before filter:")
print(df['Embarked'].value_counts())
print()
print("Passengers per port after filter (only groups with > 100 kept):")
print(big_ports['Embarked'].value_counts())
```

#### **Q11.2 Group `df` by `Pclass` and keep only the groups where the average `Fare` is greater than 20.**

---

```python
costly_classes = df.groupby('Pclass').filter(lambda g: g['Fare'].mean() > 20)

print("Average Fare per class:")
print(df.groupby('Pclass')['Fare'].mean())
print()
print("Classes kept after filter:", sorted(costly_classes['Pclass'].unique().tolist()))
costly_classes.head()
```

# 🐞 **12. Common Indexing Syntax Errors (Debug It)**

### _These snippets look almost right but throw an error or give a wrong result. Spot the mistake, explain it in a comment, and write the corrected line._

---


#### **Q12.1 This line is meant to select the `Age` and `Fare` columns together but throws an error. Find the bug and fix it: `df['Age', 'Fare']`**

---

```python
# Buggy:  df['Age', 'Fare']
# Bug: 'Age', 'Fare' inside single brackets is treated as ONE key, the tuple ('Age', 'Fare').
#      No column has that name, so pandas raises a KeyError.
# Fix: pass a LIST of column names, which needs double brackets.

df[['Age', 'Fare']]
```

#### **Q12.2 This line is meant to select rows 2 to 5 using `.iloc` but only returns 3 rows instead of 4. Explain why, and write the version that actually returns 4 rows: `df.iloc[2:5]`**

---

```python
# Buggy:  df.iloc[2:5]
# Why: .iloc slicing works like Python slicing, where the end position is EXCLUSIVE.
#      2:5 gives positions 2, 3, 4, which is only 3 rows.
# Fix: to include position 5, set the end to 6.

df.iloc[2:6]
```

#### **C1. Using `df`, build a summary table (via `pivot_table` or `groupby` + named `agg()`) showing, for each `Pclass`: average `Fare`, average `Age`, and total number of passengers — sorted from highest to lowest average `Fare`.**

#### **C2. Drop rows with a missing `Age` from `df`, create a new column `age_group` using conditional assignment (`'Child'` if `Age < 18` else `'Adult'`), then group by `Pclass` and `age_group` to find the survival rate for each combination.**

---

```python
df_age = df.dropna(subset=['Age']).copy()

df_age['age_group'] = np.where(df_age['Age'] < 18, 'Child', 'Adult')

survival_rate = df_age.groupby(['Pclass', 'age_group'])['Survived'].mean()
survival_rate
```

#### **C3. Take a random sample of 200 passengers from `df` using `.sample(n=200, random_state=42)`. From this sample, filter passengers who are male and did not survive. Then group this filtered data by `Pclass`, and using named aggregation, create `avg_fare` and `passenger_count`.**

---

```python
sample_200 = df.sample(n=200, random_state=42)

male_not_survived = sample_200[(sample_200['Sex'] == 'male') & (sample_200['Survived'] == 0)]

c3_summary = male_not_survived.groupby('Pclass').agg(
    avg_fare=('Fare', 'mean'),
    passenger_count=('PassengerId', 'count')
)
c3_summary
```
