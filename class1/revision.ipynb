```python
def validate_username(username):
    clean_username = username.strip()
    if len(clean_username) >= 5 and clean_username.isalnum():
        return clean_username.lower()
    return "Invalid"

print(repr(validate_username(" CodeMaster99 ")))
print(repr(validate_username(" Python_Dev ")))  
```

```python
def reverse_spy_code(secret_code):
    words = secret_code.split("-")
    words.reverse()
    return " ".join(words).upper()

secret_code = "charlie-bravo-alpha"
print(repr(reverse_spy_code(secret_code)))
```

```python
def clean_database_names(raw_names):
    names = raw_names.split(";")
    cleaned_names = [name.strip().title() for name in names if name.strip()]
    return ", ".join(cleaned_names)

raw_names = " john doe ; jAnE sMiTh; ALAN tUrInG "
print(repr(clean_database_names(raw_names)))
```

```python
def clean_sentence(text):
    words = text.split()
    return " ".join(words).capitalize()

text = " python is a very powerful language "
print(repr(clean_sentence(text)))
```

```python
def is_palindrome_sentence(sentence):
    clean_chars = [char.lower() for char in sentence if char.isalnum()]
    clean_str = "".join(clean_chars)
    return clean_str == clean_str[::-1]

print(is_palindrome_sentence("Was it a car or a cat I saw?"))
print(is_palindrome_sentence("Python is fun"))               
```

```python
def extract_hashtags(post):
    words = post.split()
    return [word.replace("#", "").lower() for word in words if word.startswith("#")]

post = "Loving the SUNSET! #Nature #Sunset_Vibes #beautiful"
print(extract_hashtags(post))
```

```python
def extract_file_extensions(filenames):
    files = filenames.split(",")
    extensions = []
    for file in files:
        file = file.strip()
        if "." in file:
            ext = file.split(".")[-1].upper()
            if ext not in extensions:
                extensions.append(ext)
    return extensions

filenames = "report.pdf, image.JPG, data.csv, notes.txt, photo.jpg"
print(extract_file_extensions(filenames))
```

```python
def extract_domain(email):
    if "@" not in email:
        return "Invalid"
    domain_part = email.split("@")[1]
    company = domain_part.split(".")[0]
    return company.title()

print(repr(extract_domain("developer_99@microsoft.com")))
print(repr(extract_domain("admin_at_website.org")))     
```

```python
def is_anagram(str1, str2):
    clean1 = str1.replace(" ", "").lower()
    clean2 = str2.replace(" ", "").lower()
    return sorted(clean1) == sorted(clean2)

str1 = "Dormitory"
str2 = "Dirty room"
print(is_anagram(str1, str2))
```

```python
def format_phone_number(raw_num):
    digits = "".join([char for char in raw_num if char.isdigit()])
    if len(digits) == 10:
        return f"({digits[:3]}) {digits[3:6]}-{digits[6:]}"
    return "Invalid Number"

print(repr(format_phone_number(" +1 (555) 123.4567 ")))
print(repr(format_phone_number("555-987-6543")))       
```

```python
def validate_sku(sku):
    cleaned = sku.strip().upper()
    parts = cleaned.split("-")
    if len(parts) == 2:
        letters, digits = parts
        if len(letters) == 3 and letters.isalpha() and len(digits) == 4 and digits.isdigit():
            return f"Valid: {cleaned}"
    return "Invalid SKU"

print(repr(validate_sku(" xYz-9876 ")))
print(repr(validate_sku("AB-12345")))  
print(repr(validate_sku("A@C-1234"))) 
```
